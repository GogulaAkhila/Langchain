Messages:

Messages are the fundamental unit of context for models in langchain.They represent the input and output of models , carrying both the content and needed to represent the state of a convseration when interacting with an LLM. messages are objects that contain:

Role:identifies the message type(eg:system,user or AI)
Content:Represents the actual content of the messages (like text, images ,audio, documents etc)
Metadata:Represents fields such as response information , messages IDs and token usage

Langchain provides a standard message type that works across all the models providers , ensuring consistent behaviour regardless of the model being called.

```python

import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()

model = init_chat_model("groq:openai/gpt-oss-120b")

model.invoke("Hello, how are you?")
# print(response.content)
```

Text Prompts are strings - ideal for straightforward generation tasks where you don't need to retain conversation history

```python
model.invoke("what is langchain")
```

Use Text Prompts when:

you have a single , standalone request

you don't need conversation history 

you want minimal code complexity

1. System Message (SystemMessage): Provides instructions to the AI model about its role, behavior, and how it should respond.

Example: "You are a helpful Python tutor."

2. Human Message (HumanMessage): Represents the input, question, or instruction provided by the user to the AI model.

Example: "What is Python?"

3. AI Message (AIMessage): Represents the response generated by the AI model. It can also contain tool calls requesting a tool to be executed.

Example: "Python is a programming language."

4. Tool Message (ToolMessage): Represents the output returned by a tool after the AI model requests its execution.

Example: "The current temperature in Boston is 28°C."

In simple terms: System = Instructions, Human = User input, AI = Model response, Tool = Tool output.

```python
#You can create an AIMessage manually in LangChain to maintain conversation history, so the LLM knows what it previously said.
#Example: Manually creating an AI message


# Import the message types from LangChain
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

# Create a list of messages to maintain conversation history
messages = [
    # Define the role and behavior of the AI assistant
    SystemMessage(content="You are a helpful coding assistant."),

    # Store the user's previous question
    HumanMessage(content="What is Python?"),

    # Manually add the AI's previous response to the conversation history
    AIMessage(content="Python is a high-level programming language."),

    # Add the user's new follow-up question
    HumanMessage(content="What is it used for?")
]

# Send the complete conversation history to the LLM
response = model.invoke(messages)

# Print the LLM's response to the latest question
print(response.content)
```

```python
#Here is a simple LangChain example using a weather tool and ToolMessage



# Import the required message types
from langchain_core.messages import HumanMessage, AIMessage, ToolMessage

# Create conversation history for a weather query
messages = [
    # User asks for the weather
    HumanMessage(content="What is the weather in Hyderabad?"),

    # AI requests the weather tool
    AIMessage(
        content="",
        tool_calls=[
            {
                "name": "get_weather",
                "args": {"location": "Hyderabad"},
                "id": "call_123",
                "type": "tool_call"
            }
        ]
    ),

    # Manually add the result returned by the weather tool
    ToolMessage(
        content="Hyderabad: 32°C, partly cloudy.",
        tool_call_id="call_123",
        name="get_weather"
    )
]

# Print each message in the conversation history
for message in messages:
    print(f"{type(message).__name__}: {message.content}")
```
